# About the Company

Olist is a Brazilian e-commerce company founded in 2015 in Curitiba, Brazil, by Tiago Dalvi. The company was created with the goal of helping small and medium-sized businesses sell their products through major online marketplaces such as Mercado Livre and Amazon. Olist acts as an intermediary platform that simplifies the process of online selling by providing retailers with tools to manage inventory, orders, logistics, payments, and customer interactions in one centralized system.

The company operates through a retail-as-a-service business model, enabling merchants to access large digital marketplaces without needing extensive technical infrastructure or individual marketplace contracts. Through services such as Olist Store and Olist Shops, sellers can improve product visibility, manage operations more efficiently, and create their own online stores.

In addition to marketplace integration, Olist offers logistics, shipping, payment, and financial solutions designed to streamline e-commerce operations and improve customer experience. Over the years, the company has experienced rapid growth and has become one of the leading e-commerce technology companies in Brazil, supporting thousands of merchants across the country.

Olist plays an important role in the Brazilian digital economy by helping small businesses compete in the growing e-commerce market. Its platform became especially relevant during the rapid expansion of online shopping in recent years.

Analyzing sales trends and customer behavior can help the company make more informed business decisions regarding product strategy, marketing efforts, customer retention, and operational improvements.

# DataFrames Relationships & Structure

The Olist dataset is composed of multiple related tables that together represent different aspects of the e-commerce business, including customers, orders, products, sellers, payments, reviews, and geolocation data.

These datasets form a relational database structure connected mainly through:

* `order_id`
* `customer_id`
* `product_id`
* `seller_id`

### Understanding these relationships is essential before performing data cleaning, merging, and analysis.
---
## Customers Dataset (`olist_customers`)

Contains customer identification and location information.

### Columns

* `customer_id`
* `customer_unique_id`
* `customer_zip_code_prefix`
* `customer_city`
* `customer_state`

### Relationship
* Connected to `olist_orders` through `customer_id`.
---
## Geolocation Dataset (`olist_geolocation`)

Contains Brazilian geolocation data associated with zip code prefixes.

### Columns

* `geolocation_zip_code_prefix`
* `geolocation_lat`
* `geolocation_lng`
* `geolocation_city`
* `geolocation_state`

### Relationship

* Can be linked to customer and seller datasets through zip code prefixes for regional analysis and map visualizations.
---

```python
import pandas as pd

olist_customer = pd.read_csv('Downloads/olist_data/olist_customers_dataset.csv')
olist_geolocation = pd.read_csv('Downloads/olist_data/olist_geolocation_dataset.csv')
olist_order_items = pd.read_csv('Downloads/olist_data/olist_order_items_dataset.csv')
olist_payments = pd.read_csv('Downloads/olist_data/olist_order_payments_dataset.csv')
olist_reviews = pd.read_csv('Downloads/olist_data/olist_order_reviews_dataset.csv')
olist_orders = pd.read_csv('Downloads/olist_data/olist_orders_dataset.csv')
olist_products = pd.read_csv('Downloads/olist_data/olist_products_dataset.csv')
olist_sellers = pd.read_csv('Downloads/olist_data/olist_sellers_dataset.csv')
olist_product_category = pd.read_csv('Downloads/olist_data/product_category_name_translation.csv')

dfs = [
    olist_customer, olist_geolocation, olist_order_items, olist_payments,
    olist_reviews, olist_orders, olist_products, olist_sellers, olist_product_category
]

for df in dfs:
    df.columns = df.columns.str.strip().str.lower()
    df.drop_duplicates(inplace=True)

date_cols_orders = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date'
]

for col in date_cols_orders:
    olist_orders[col] = pd.to_datetime(olist_orders[col], errors='coerce')

olist_order_items['shipping_limit_date'] = pd.to_datetime(
    olist_order_items['shipping_limit_date'], errors='coerce'
)

olist_reviews['review_creation_date'] = pd.to_datetime(
    olist_reviews['review_creation_date'], errors='coerce'
)
olist_reviews['review_answer_timestamp'] = pd.to_datetime(
    olist_reviews['review_answer_timestamp'], errors='coerce'
)

analysis = (
    olist_order_items
    .merge(olist_orders, on='order_id', how='left')
    .merge(olist_products, on='product_id', how='left')
    .merge(olist_product_category, on='product_category_name', how='left')
)

analysis['category_name'] = analysis['product_category_name_english'].fillna(
    analysis['product_category_name']
)

analysis['revenue'] = analysis['price'].fillna(0) + analysis['freight_value'].fillna(0)
analysis['delivery_days'] = (
    analysis['order_delivered_customer_date'] - analysis['order_purchase_timestamp']
).dt.days
analysis['late_delivery'] = (
    analysis['order_delivered_customer_date'] > analysis['order_estimated_delivery_date']
)
analysis['order_month'] = analysis['order_purchase_timestamp'].dt.to_period('M').astype(str)

delivered = analysis[analysis['order_status'] == 'delivered'].copy()

category_summary = (
    delivered.groupby('category_name', dropna=False)
    .agg(
        item_count=('order_id', 'count'),
        unique_orders=('order_id', 'nunique'),
        total_revenue=('revenue', 'sum'),
        avg_price=('price', 'mean'),
        avg_freight=('freight_value', 'mean'),
        avg_delivery_days=('delivery_days', 'mean'),
        late_delivery_rate=('late_delivery', 'mean')
    )
    .reset_index()
    .sort_values(['total_revenue', 'unique_orders'], ascending=False)
)

product_summary = (
    delivered.groupby(['product_id', 'category_name'], dropna=False)
    .agg(
        item_count=('order_id', 'count'),
        unique_orders=('order_id', 'nunique'),
        total_revenue=('revenue', 'sum'),
        avg_price=('price', 'mean')
    )
    .reset_index()
    .sort_values(['item_count', 'total_revenue'], ascending=False)
)

print("Analysis table:", analysis.shape)
print("Category summary:", category_summary.shape)
print("Top categories:")
print(category_summary.head(10))

print("Top products:")
print(product_summary.head(10))
```

```python
brl_to_eur = 0.18  # converting to euros

# 1) create EUR columns on analysis
analysis['price_eur'] = analysis['price'] * brl_to_eur
analysis['freight_eur'] = analysis['freight_value'] * brl_to_eur
analysis['revenue_eur'] = analysis['revenue'] * brl_to_eur

# 2) then create delivered from analysis
delivered = analysis[analysis['order_status'] == 'delivered'].copy()
```
